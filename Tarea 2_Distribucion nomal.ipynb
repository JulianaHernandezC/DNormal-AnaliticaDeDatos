Tipificar

```python
# Tipificar los datos (escalado a media 0 y desviación estándar 1)
df['Valores Tipificados'] = (df['Valores'] - df['Valores'].mean()) / df['Valores'].std()
df.head()
```

#**Graficos**

```python
# Graficar la distribución normal
plt.figure(figsize=(10, 6))
plt.hist(df['Valores'], bins=30, density=True, alpha=0.6, color='pink', edgecolor='black')
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = norm.pdf(x, df['Valores'].mean(), df['Valores'].std())
plt.plot(x, p, 'k', linewidth=2)
plt.title('Distribución Normal')
plt.xlabel('Valores')
plt.ylabel('Densidad')
plt.show()
```

```python
# Graficar la distribución tipificada
plt.figure(figsize=(10, 6))
plt.hist(df['Valores Tipificados'], bins=30, density=True, alpha=0.6, color='yellow', edgecolor='black')
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = norm.pdf(x, df['Valores Tipificados'].mean(), df['Valores Tipificados'].std())
plt.plot(x, p, 'k', linewidth=2)
plt.title('Distribución Tipificada')
plt.xlabel('Valores Tipificados')
plt.ylabel('Densidad')
plt.show()
```

# **Pruebas**

```python
#valores iniciales
valor_original = 60
media = df['Valores'].mean()
desviación_estandar = df['Valores'].std()

# Valor Tipificado
valor_tipificado = (valor_original - media) / desviación_estandar
# Calcular la probabilidad de ese valor en la distribución original
probabilidad = norm.cdf(valor_original, media, desviación_estandar)
```

```python
# Calcular el valor tipificado que corresponde a la misma probabilidad
valor_tipificado_correspondiente = norm.ppf(probabilidad)
print(valor_tipificado_correspondiente)
```

Probabilidad P(Z<= 1)

```python
# Calcular una cierta probabilidad para la distribución tipificada (ejemplo: P(Z <= 1.0))
probabilidad_tipificada = norm.cdf(valor_tipificado_correspondiente) # la media = 0 y la desviacion estandar = 1
print(f'P(Z <= 1.0): {probabilidad_tipificada:.4f}')
```

Probabilidad P(X <= 60)

```python
# Calcular una cierta probabilidad (ejemplo: P(X <= 60) para la distribución original)
probabilidad = norm.cdf(valor_original, df['Valores'].mean(), df['Valores'].std())
print(f'P(X <= 60): {probabilidad:.4f}')
```
